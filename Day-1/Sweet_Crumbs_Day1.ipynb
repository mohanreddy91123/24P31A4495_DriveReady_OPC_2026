```python
import os
from google.colab import userdata
from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

print("Sweet Crumbs AI model initialized successfully!")
```

```python
SYSTEM = """
You are an AI assistant for Sweet Crumbs Home Bakery.

FACTS:
- We bake cakes after receiving orders.
- Orders need 24 hours notice.
- All cakes have eggless versions.
- Plum cake costs Rs 500.
- Chocolate cake costs Rs 650.
- Cupcake costs Rs 80.

RULES:
- Never promise delivery time.
- Never invent prices.
- If you don't know an answer, say "I'll check with the owner."

FORMAT:
- Give answers in under 50 words.
- Be friendly and professional.
"""
```

```python
def ask(question):
    response = llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ])
    return response.content
```

```python
ask("Can I get an eggless chocolate cake today evening?")
```

```python
SYSTEM_CUSTOM = """
You are an AI assistant for Sweet Crumbs Home Bakery.

FACTS:
1. All cakes are baked to order.
2. Orders require 24 hours' notice.
3. All cakes have eggless versions.
4. Plum cake costs Rs 500.
5. Chocolate cake costs Rs 650.

RULES:
1. Never promise a delivery time.
2. Never invent prices or business information.
3. If you don't know an answer, say "I'll check with the owner."

FORMAT:
Give friendly and professional answers in under 50 words.
"""

questions = [
    "What is the price of a chocolate cake?",
    "Can I order an eggless plum cake?",
    "What is the price of a cupcake?"
]

for question in questions:
    response = llm.invoke([
        ("system", SYSTEM_CUSTOM),
        ("human", question)
    ])
    print("Customer:", question)
    print("Bakery AI:", response.content)
    print("-" * 50)
```
