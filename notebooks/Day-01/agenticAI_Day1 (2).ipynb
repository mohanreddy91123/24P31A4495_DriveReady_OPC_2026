```python
import os
from google.colab import userdata
from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

print("AutoCare AI model initialized successfully!")
```

```python
response = llm.invoke("Hello! What can you do?")
print(response.content)
```

```python
SYSTEM = """
You are AutoCare AI, an intelligent automobile service assistant.

YOUR ROLE:
Help customers with vehicle maintenance,
service information and appointment assistance.

SERVICES:
1. Car, bike and scooter maintenance.
2. Engine oil and tyre maintenance.
3. Battery and brake maintenance.
4. General vehicle service information.
5. Service appointment assistance.

RULES:
1. Always respond politely.
2. Ask for vehicle model and year when necessary.
3. Never invent service prices.
4. Never claim a confirmed mechanical diagnosis.
5. For serious vehicle problems, recommend a qualified mechanic.
6. Never guarantee appointment availability without checking.
7. If you don't know something, say so.

RESPONSE FORMAT:
Use simple English.
Give clear and helpful answers.
Keep responses under 100 words.
"""

print("AutoCare AI system prompt created!")
```

```python
def ask(question):
    response = llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ])
    return response.content

print("AutoCare AI is ready to answer customers!")
```

```python
print(ask("When should I change my car engine oil?"))
```

```python
print(ask("What maintenance does a Honda Activa need?"))
```

```python
print(ask("My car brakes are making a strange noise. What should I do?"))
```

```python
print(ask("I want to book a service appointment for my Hyundai i20. What details do you need?"))
```

```python
print(ask("What is the exact service price for my car?"))
```
