```python
!pip install -q langchain

from langchain.tools import tool

MENU = {
    "plum cake": 500,
    "chocolate cake": 650,
    "cupcake": 80
}

HOURS = {
    "monday": "9 AM to 8 PM",
    "tuesday": "9 AM to 8 PM",
    "wednesday": "9 AM to 8 PM",
    "thursday": "9 AM to 8 PM",
    "friday": "9 AM to 8 PM",
    "saturday": "9 AM to 8 PM",
    "sunday": "10 AM to 6 PM"
}


@tool
def get_price(item: str, qty: int = 1) -> str:
    """Get the total price for a bakery menu item and quantity."""

    item = item.lower().strip()

    if item not in MENU:
        return f"Item not available. Menu: {list(MENU.keys())}"

    if qty <= 0:
        return "Quantity must be greater than zero."

    total = MENU[item] * qty

    return f"{qty} x {item} = Rs. {total}"


@tool
def is_open(day: str) -> str:
    """Check the bakery opening hours for a particular day."""

    day = day.lower().strip()

    if day not in HOURS:
        return "Invalid day."

    return f"On {day}, bakery hours are {HOURS[day]}."


print(get_price.invoke({"item": "chocolate cake", "qty": 2}))
print(is_open.invoke({"day": "sunday"}))
```

```python
from typing import Literal
from pydantic import BaseModel, Field

class Order(BaseModel):
    name: str = Field(description="Customer name")

    item: Literal[
        "plum cake",
        "chocolate cake",
        "cupcake"
    ]

    qty: int = Field(ge=1, description="Order quantity")

    date: str = Field(description="Pickup date")

    eggless: bool = Field(description="Whether eggless is required")


print("Order model created successfully!")
```

```python
extractor = llm.with_structured_output(Order)

order = extractor.invoke(
    "Hi, I'm Ravi. I need 2 eggless chocolate cakes for 24 December."
)

print(order)
print("Total price:", order.qty * MENU[order.item])
```

```python
from langchain.tools import tool

@tool
def log_order(
    name: str,
    item: str,
    qty: int,
    date: str,
    eggless: bool
) -> str:
    """Save a confirmed bakery order into Google Sheets."""

    item = item.lower().strip()

    if item not in MENU:
        return "Item not available."

    if qty <= 0:
        return "Quantity must be greater than zero."

    total = MENU[item] * qty

    ws.append_row([
        name,
        item,
        qty,
        date,
        "Yes" if eggless else "No",
        total
    ])

    return f"Order saved! Total amount: Rs. {total}"


print("Order logging tool created successfully!")
```

```python
from pydantic import ValidationError

invalid_orders = [
    {"name": "Ravi", "item": "donut", "qty": 2, "date": "24 Dec", "eggless": True},

    {"name": "Priya", "item": "plum cake", "qty": 0, "date": "25 Dec", "eggless": False},

    {"name": "Arjun", "item": "cupcake", "qty": -5, "date": "26 Dec", "eggless": True},

    {"name": "Kiran", "item": "chocolate cake", "qty": "two", "date": "27 Dec", "eggless": True},

    {"name": "", "item": "plum cake", "qty": 2, "date": "28 Dec", "eggless": False},

    {"name": "Sneha", "item": "cupcake", "qty": 2, "date": "29 Dec", "eggless": "maybe"}
]

for i, data in enumerate(invalid_orders, 1):
    try:
        order = Order(**data)
        print(f"TEST {i}: ACCEPTED", order)

    except ValidationError as e:
        print(f"TEST {i}: REJECTED")
        print(e.errors())
```
